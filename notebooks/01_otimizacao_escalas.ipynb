# Otimizacao de Escalas e Alocacao de Pessoal

Demanda simulada por turno (7 dias, 3 turnos), otimizada via programacao inteira (PuLP) respeitando o limite de horas semanais da CLT. Detalhes em `data/README.md`.

```python
import sys
sys.path.append('..')
import seaborn as sns
import matplotlib.pyplot as plt
from src.data.load_data import pipeline_completo
from src.models.escala import otimizar_escala

df = pipeline_completo()
df
```

```python
resultado = otimizar_escala(df, n_funcionarios_pool=25)
print('status:', resultado['status'])
print('total funcionarios-turno na semana:', resultado['total_funcionarios_turno'])
resultado['escala']
```

```python
pivot = resultado['escala'].pivot(index='turno', columns='dia', values='funcionarios_alocados')
pivot = pivot[['Seg','Ter','Qua','Qui','Sex','Sab','Dom']]
fig, ax = plt.subplots(figsize=(10,5))
sns.heatmap(pivot, annot=True, fmt='d', cmap='Blues', ax=ax)
ax.set_title('Funcionarios alocados por turno (escala otimizada)')
plt.tight_layout()
plt.savefig('../reports/figures/escala_otimizada.png', dpi=150)
plt.show()
```

## Conclusoes

O solver encontra a escala de menor custo (59 funcionarios-turno na semana) que cobre toda a demanda esperada, respeitando o teto de 44h semanais por funcionario aplicado ao pool de 25 pessoas disponiveis. Fins de semana e o turno da manha concentram a maior alocacao, coerente com o pico de demanda simulado.
